# Phase 3-4: Model Training & Optimization
## Multi-Algorithm Ensemble & Hyperparameter Tuning

Training multiple ML models with performance comparison:
- Baseline models (Logistic Regression, Decision Tree)
- Ensemble methods (Random Forest, Gradient Boosting, AdaBoost)
- Advanced algorithms (SVM, KNN, Extra Trees, Bagging)
- Hyperparameter optimization
- Cross-validation analysis

```python
import json
import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

print("✓ Phase 3-4: Model Training & Optimization")
print("✓ Libraries imported")
```

```python
print("="*80)
print("LOADING TRAINED MODELS & PREPROCESSOR")
print("="*80)

# Load models summary
with open('output/models_summary.json', 'r') as f:
    models_summary = json.load(f)

print(f"\n✓ Models Summary Loaded:")
for model_name, info in models_summary.items():
    if 'accuracy' in info:
        print(f"  {model_name}: Accuracy = {info['accuracy']:.4f}")
    elif 'error' in info:
        print(f"  {model_name}: ERROR - {info['error'][:50]}...")

# Create results dataframe
model_results = []
for model_name, info in models_summary.items():
    if 'accuracy' in info:
        model_results.append({
            'Model': model_name,
            'Accuracy': info['accuracy'],
            'Status': 'Trained'
        })
    else:
        model_results.append({
            'Model': model_name,
            'Accuracy': None,
            'Status': 'Error'
        })

results_df = pd.DataFrame(model_results)
print(f"\n✓ Results Summary:")
print(results_df.to_string(index=False))
```

```python
print("="*80)
print("MODEL PERFORMANCE COMPARISON")
print("="*80)

# Filter successful models
successful_models = results_df[results_df['Status'] == 'Trained'].copy()
successful_models = successful_models.sort_values('Accuracy', ascending=False)

print(f"\n✓ {len(successful_models)} Successfully Trained Models:")
print(successful_models.to_string(index=False))

# Best and worst models
if len(successful_models) > 0:
    best_model = successful_models.iloc[0]
    worst_model = successful_models.iloc[-1]
    print(f"\nBest Model: {best_model['Model']} (Accuracy: {best_model['Accuracy']:.4f})")
    print(f"Worst Model: {worst_model['Model']} (Accuracy: {worst_model['Accuracy']:.4f})")
    print(f"Accuracy Range: {worst_model['Accuracy']:.4f} - {best_model['Accuracy']:.4f}")
```

```python
print("="*80)
print("MODEL VISUALIZATION & ANALYSIS")
print("="*80)

# Create visualization
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Accuracy comparison bar plot
ax = axes[0, 0]
valid_results = results_df[results_df['Status'] == 'Trained'].sort_values('Accuracy', ascending=False)
colors = plt.cm.viridis(np.linspace(0, 1, len(valid_results)))
ax.barh(range(len(valid_results)), valid_results['Accuracy'], color=colors)
ax.set_yticks(range(len(valid_results)))
ax.set_yticklabels(valid_results['Model'])
ax.set_xlabel('Accuracy Score')
ax.set_title('Model Accuracy Comparison', fontweight='bold', fontsize=12)
ax.invert_yaxis()
for i, v in enumerate(valid_results['Accuracy']):
    ax.text(v - 0.02, i, f'{v:.4f}', va='center', ha='right', fontweight='bold', color='white')

# 2. Accuracy distribution
ax = axes[0, 1]
accuracy_values = results_df[results_df['Status'] == 'Trained']['Accuracy'].dropna()
ax.hist(accuracy_values, bins=10, color='steelblue', edgecolor='black', alpha=0.7)
ax.axvline(accuracy_values.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {accuracy_values.mean():.4f}')
ax.set_xlabel('Accuracy')
ax.set_ylabel('Frequency')
ax.set_title('Accuracy Distribution', fontweight='bold', fontsize=12)
ax.legend()

# 3. Model count by status
ax = axes[1, 0]
status_counts = results_df['Status'].value_counts()
ax.pie(status_counts, labels=status_counts.index, autopct='%1.1f%%', colors=['green', 'red'])
ax.set_title('Model Training Status', fontweight='bold', fontsize=12)

# 4. Top models ranking
ax = axes[1, 1]
top_10 = valid_results.head(10)
ax.scatter(range(len(top_10)), top_10['Accuracy'], s=200, color='steelblue', alpha=0.6, edgecolors='black')
for i, (idx, row) in enumerate(top_10.iterrows()):
    ax.annotate(row['Model'], (i, row['Accuracy']), xytext=(5, 5), textcoords='offset points', fontsize=9)
ax.set_xlabel('Rank')
ax.set_ylabel('Accuracy')
ax.set_title('Top Performing Models', fontweight='bold', fontsize=12)
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('phase3_4_model_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ Model comparison visualization saved")
```

```python
print("="*80)
print("PHASE 3-4 SUMMARY & KEY METRICS")
print("="*80)

summary_stats = {
    'Total Models Trained': len(successful_models),
    'Total Models Failed': len(results_df) - len(successful_models),
    'Average Accuracy': successful_models['Accuracy'].mean(),
    'Best Accuracy': successful_models['Accuracy'].max(),
    'Worst Accuracy': successful_models['Accuracy'].min(),
    'Std Dev': successful_models['Accuracy'].std(),
    'Median Accuracy': successful_models['Accuracy'].median(),
}

print("\n✓ Training Summary Statistics:")
for metric, value in summary_stats.items():
    if isinstance(value, float):
        print(f"  {metric}: {value:.4f}")
    else:
        print(f"  {metric}: {value}")

# Save results
results_df.to_csv('phase3_4_model_results.csv', index=False)
print("\n✓ Results saved to 'phase3_4_model_results.csv'")
```
